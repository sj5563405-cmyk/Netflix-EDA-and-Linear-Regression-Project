###  Step 3: Clean Missing Values

```python
df['director'].fillna('Unknown',inplace = True)
df['cast'].fillna('No Cast Info',inplace = True)
df['country'].fillna(df['country'].mode()[0],inplace=True)
df['rating'].fillna(df['rating'].mode()[0],inplace= True)
df['duration'].fillna(df['duration'].mode()[0],inplace = True)
df['date_added'].fillna(df['date_added'].mode()[0],inplace= True)
# Confirm cleaning
print("\n Missing values after cleaning :\n", df.isnull().sum())
```

##  Step 5: Split Data for Model Building

```python
from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import LabelEncoder 
```

```python
label = LabelEncoder()
df['type'] = label.fit_transform(df['type'])
df['rating'] = label.fit_transform(df['rating'])
df['country'] = label.fit_transform(df['country'])

x = df[['type','rating','country']]
y = df['release_year']

x_train, x_test, y_train, y_test = train_test_split(x,y,  test_size =0.2, random_state = 42)
```

```python
print("Train Shape:" , x_train.shape)
print("Test Shape:",   x_test.shape)
print("Train Shape:",  y_train.shape)
print("Test Shape:",   y_test.shape)
```

###  Step 6: Build and Evaluate Linear Regression Model

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
df_model  = df.copy()
label  = LabelEncoder ()

for col in ['type','rating','country']:
    df_model[col] = label.fit_transform(df_model[col])  
```

```python
X = df_model[['type','rating','country']]
Y = df_model['release_year']
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X, Y, test_size =0.2,random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
r2 = r2_score(Y_test,Y_pred)
mae = mean_absolute_error(Y_test,Y_pred)
mse = mean_squared_error(Y_test,Y_pred)
print(" Linear Regression Model Performance:")
print(f"R² Score: {r2:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
```

```python
plt.figure(figsize=(8,5))
plt.scatter(y_test, y_pred, alpha=0.6)
plt.xlabel("Actual Release Year")
plt.ylabel("Predicted Release Year")
plt.title("Actual vs Predicted Release Year")
plt.show()
```
